# Why OOP 

- Object-Oriented Programming (OOP) is used in Python because it allows developers to bundle data and the specific actions that operate on that data in organized , reusable structures called classes and objects.

- This structural approach makes large application much easier to maintain , scale , and debug by breaking the code down into seperate , self - contained building blocks.




# The 4 Pillars of Python OOP

Python implementas OOP through four core principles :

1. **Inheritance** : Lets a new class inherit variable and methods from an existing class to reduce duplicate code (e.g. , *car* class can inherit from a general *Vehicle* class)

2. **Polymorphism** : Allows different classes to have methods with the exact same name but different behaviours (e.g. both a *Dog* and a *Cat* class can have a *.make_sound()* method) 

3. **Encapsulation**: Restricts direct outside access to an objects's internal state using naming conventions like ( *_protected* or *_private* prefixes) to keep data secure and controlled

4. **Abstraction**: Hides complex internal excecution details and exposes only the essential operational features via tools like python*abc* module.

# Key benefits over Standard functions

- **DRY (Don't Repeat Yourself)**: You can write code once in a template (class) and seamlessly create thousands of variations (objects).

- **Real-World Modeling**: It is much easier to program systems like video game characters, bank accounts, or e-commerce products when you can model them as physical "objects" with specific attributes and behaviors.

- **Pythonic Uniformity**: In Python, everything is already an object behind the scenes (strings, lists, dictionaries). Utilizing OOP allows you to write custom structures that align naturally with Python's built-in ecosystem.

## **Step 1: The Evolution of Programming**

Programming evolved because software became increasingly complex.

Stage 1: Procedural Programming

In procedural programming, we organize code around functions. Data is stored separately, and functions operate on that data.

Example:
```
employee_name = "Alice"
employee_salary = 50000

def apply_raise(salary, percentage):
    return salary * (1 + percentage / 100)

employee_salary = apply_raise(employee_salary, 10)
```

This works well for small scripts.

Where does it start to break?

- Imagine an HR system with:

50,000 employees
Payroll
Attendance
Performance reviews
Promotions
Leave management

You now have hundreds of variables and dozens of functions. Keeping the correct data associated with the correct employee becomes increasingly difficult.

**Step 2: The Core Problem**

Procedural programming separates data and behavior.

Employee Data
-------------
- Name
- Salary
- Department

↓

Functions
----------
- pay_salary()
- promote()
- transfer()
- calculate_tax()

Anyone can accidentally pass the wrong employee data into a function. The program grows, duplication increases, and maintenance becomes difficult.

**Step 3: OOP's Solution**

OOP says:

Keep related state and behavior together.

Instead of separate data and functions, create an object that owns both.
```
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def apply_raise(self, percentage):
        self.salary *= (1 + percentage / 100)

alice = Employee("Alice", 50000)
alice.apply_raise(10)

print(alice.salary)
```
Now the Employee object is responsible for maintaining its own state.

**Step 4: A Robotics Example**

Suppose you're building an autonomous inspection robot.

A procedural approach might look like:
```
camera_resolution = "1080p"
camera_status = "Ready"

motor_speed = 0

def capture_image():
    print("Capturing image")

def set_motor_speed(speed):
    global motor_speed
    motor_speed = speed
```
As you add more cameras, motors, and sensors, the number of global variables and functions grows rapidly.

With OOP:
```
class Camera:
    def __init__(self, resolution):
        self.resolution = resolution
        self.status = "Ready"

    def capture(self):
        print(f"Capturing image at {self.resolution}")

class Motor:
    def __init__(self):
        self.speed = 0

    def set_speed(self, speed):
        self.speed = speed

front_camera = Camera("1080p")
left_motor = Motor()

front_camera.capture()
left_motor.set_speed(40)

Each object manages its own data, making the system easier to extend and test.

Step 5: Why This Matters in Machine Learning

Consider scikit-learn:

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

rf = RandomForestClassifier()
svm = SVC()

rf.fit(X_train, y_train)
svm.fit(X_train, y_train)

rf.predict(X_test)
svm.predict(X_test)
```
