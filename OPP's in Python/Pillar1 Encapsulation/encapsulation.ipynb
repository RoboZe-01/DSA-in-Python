## **How Encapsulation works in Python**

Unlike languages like Java or C++ , Pyhon does not have strict keywords enforement ( *Public* , *Private* ,*protected*). Instead , it relies on naming conventions to signal data protection levels:

- **Public Members**: Accessible from anywhere . This is the default behavior in Python 
- **Protected Members** ( **_** ) : Prefixed  with a single underscore . It serves as a visual warning to developers that the attribute shouldn't be accessed outside the calss or its subclasses , though Python won't actively stop you 
- **Private Members** (**__**) : Prefixed with a double underscore . Python triggers a process call Name Mangling . Internally renaming the variable ( e.g. , *__balance* becomens *_Classname__balance*) to hinder direct outside acces avoid naming collisions

## **Code Example : Implementing Encapsulation**
The standard way to provide controlled access to private variables is through **Getters and Setters**

- **Getter** : Get / access the data
- **Setter**  : Modify the data

```python
class BankAccount : 
    def __init__(self , owner , balance):
        self.owner = owner                 # Public attribute 
        self.__balance = balance           # Private attribute


        # Getter method to read private data securely 

    def get_balance(self):
        return self.__balance

    # Setter method to modify private data with validation rules
    def deposit(self,amount):
        if amount > 0 :
            self.__balance+=amount
        else : 
            print("Error: Deposit amount must be positive.")   

# Using the class
account = BankAccount("subha",10000000)

print(account.owner)   # Output : Alice (Accessible)
# print(account.__balance)   # Throws Attribute (Protected by name Name Mangling )

# Safe intercation through methods
account.deposit(9999)
print(account.get_balance())

```
