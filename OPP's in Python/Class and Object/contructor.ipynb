# What is constructor 

- If we talk about a function we can ask the user using parameters , but in class we can't have paramets for that we use constructor

```python
class Bags(material , zips ):
    print( "class")
```

Thus we can see here a class can't have parameters directly , for that in python we use constructor `__init__()`

-  A constructor is a method that runs automatically when we call a class and this constructor function will target the objects location 

```python
class Bags:
    def __init__(self,material,zips,pockets):
        print("How are you")

reebok = Bags()        
```

- If we have some parameters in the `__init__()` we have to provide them in the initialization of the object phase 
- `self` will have the location of the variable 

```python
class Bags:
    def __init__(self,material,zips,pockets):
        self.material = material
        self.zips = zips
        self.pockets = pockets


# Object creation 
reebok = Bags("Leather", 3,2)
```
