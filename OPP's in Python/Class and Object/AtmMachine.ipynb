```python
# Initialize the class

class AtmMachine:

    # data property
    def __init__(self):
        self.pin = ""
        self.balance = 0
        self.menu()
    

    # behaviour property
    def menu(self):
        user_input = input(
            """
          Hello ,how can we help you ?
          1. Press 1 to check the balance
          2. Press 2 to withdraw
          3. Press 3 to deposit
          4. Press 4 to create pin
          5. Press 4 to change the pin

          Press anything to exit (can press enter)
            """
        )

        if user_input =="1":
            # Check balance code
            self.check_balance()
            pass

        elif user_input=="2":
            # money withdraw code
            pass
        elif user_input=="3":
            # Deposit code
            self.deposit_balance()
            pass
        elif user_input=="4":
            self.create_pin()
            pass
        elif user_input=="5":
            self.update_pin()
            pass
        else:
            exit()



     ## Creating functionality for the project

     # Create pin function       
    def create_pin(self):
        user_pin = int(input("Create your pin (numbers only) : "))
        self.pin = user_pin

        user_balance = int(input("Enter balace : "))
        self.balance = user_balance
        print("Pin created successfully ")
        self.menu()        
    

    # Update Pin function
    def update_pin(self):
        user_input = int(input("Enter your old pin : "))

        if user_input==self.pin:
            new_pin = int(input("Set new Pin : "))
            self.pin = new_pin
            print("Pin changes successfully : ")
            self.menu()
        else:
            print("Incorrect Pin !! ")    
            self.menu()


    # Function to deposit the money 
    def deposit_balance(self):
        user_pin = int("Enter the pin : ")
        
        while True: 
                if user_pin==self.pin:
                    money_depo = int(input("Enter the amount to deposit : "))
                    self.balance = self.balance+money_depo
                    print(f"Money deposited , current balance is {self.balance}")
                    break
                else:
                    print("Incorrect pin !! ")   
    
    
    # Check balance function
    def check_balance(self):
        user_pin = int(input("Enter your pin to check balance : "))
        while True:
            if user_pin == self.pin:
                print(self.balance)
            else:
                print("Invalid Pin !! ")    


    # Money withdraw function 
         
```
