`Float` 

In Python, the float data type represents real numbers containing a fractional component or a decimal point. Under the hood, Python implements floats as 64-bit double-precision values (following the IEEE-754 standard), providing roughly 15 to 17 digits of precision

## 1. Creating floats

```python
# Literal declaration 
x = 10.5

# Scientific notation (1.23 x 10^4 = 12300 )
y = 1.23e4

# Standard division automatically makes float

z = 10/2
```

```python
print(x)
print(y)
print(z)
```

## 2. The float() Type Conversion Function
float() function converts integers or compatible string representations into floating-point numbers.
- Passing normal string like "abc" raises `ValueError`

```python
# Converting an integer
print(float(5))

# converting numerical string
print(float("3.4"))
print(float(" 9.5"))


```
