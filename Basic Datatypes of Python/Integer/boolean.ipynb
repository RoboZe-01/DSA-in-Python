# Boolean 
- Boolean data type (bool) represents the truth value of an expression. It has only two built-in constant values: `True` and `False`.
- True behaves numerically like 1, and False behaves like 0.

### Truthy vs. Falsy Values
Python can evaluate any object in a Boolean context (like an if statement) without explicitly converting it. Objects are classified as either "Truthy" or "Falsy":
• **Falsy Values**: Automatically evaluate to False.
	• None
	• False
	• Numeric zeros (0, 0.0)
	• Empty sequences ('', [], ())
	• Empty mappings ({})
• **Truthy Values**: Any non-empty string, non-zero number, or populated collection evaluates to True.
You can check an object's inherent truth value using the built-in bool() function.
